```python
#series - one dimentional array, like a column in table

numbers = ['1','2','3']

abc = pd.Series(numbers)
print(abc)

# passing series row number
print([0])

# indexing in series

abc = pd.Series(numbers, index = ["x", "y", "z"])
print(abc["x"])
```

```python
# DataFrame - two dimentional array or two dimentional array/ table with columns and rows

Activity = {
  "Price": [420, 380, 390],
  "time": [50, 40, 45],
    "games": ["Playstation","Basketball","Football"]
}

#loading data into a DataFrame object:
df = pd.DataFrame(Activity)

print(df) #only print first 5 rows and last 5 rows and header
```

```python
#locate row -- loc()
print(df.loc[1])

# returning series of values in DataFrame
print(df.loc[[1,2]])
```

```python
# load python dictionaries from into python data frame
# python dictionary = JSON
# dictionary means a built-in data structure that stores data in key: value pairs, mutable(changeable)

dataset = {
  "Duration":{
    "0":60,
    "1":60,
    "2":60,
    "3":45,
    "4":45,
    "5":60
  },
  "Pulse":{
    "0":110,
    "1":117,
    "2":103,
    "3":109,
    "4":117,
    "5":102
  },
  "Maxpulse":{
    "0":130,
    "1":145,
    "2":135,
    "3":175,
    "4":148,
    "5":127
  },
  "Calories":{
    "0":409,
    "1":479,
    "2":340,
    "3":282,
    "4":406,
    "5":300
  }
}

df = pd.DataFrame(dataset)

print(df) 
```
